# Modul 1: Die Differentialgleichung als Vektorfeld

---

## Der Paradigmenwechsel

Was ist eine Differentialgleichung *wirklich*?

Die klassische Antwort: Ein Rätsel. Gegeben $\dot{x} = f(x)$, finde $x(t)$.

Diese Perspektive führt zu Techniken: Trennung der Variablen, Variation der Konstanten, Laplace-Transformation... Man lernt Rezepte, ohne das Wesen zu verstehen.

**Die geometrische Perspektive** kehrt die Frage um:

> Eine DGL *ist* kein Rätsel, das man löst. Sie *ist* eine Struktur - ein Vektorfeld, das einen Fluss im Raum definiert.

Die Lösung $x(t)$ ist dann nur eine Trajektorie in diesem Feld - eine Kurve, die überall tangential zu den Vektoren verläuft.

Dieser Perspektivwechsel ist nicht nur ästhetisch. Er erlaubt uns, das **qualitative Verhalten** eines Systems zu verstehen, ohne je eine einzige Gleichung zu lösen. Und er öffnet die Tür zu Phänomenen wie Chaos, die in der Lösungsformel-Welt unsichtbar bleiben.

## 1.1 Was bedeutet $\dot{x} = f(x)$?

Betrachten wir die einfachste autonome DGL:

$$\dot{x} = f(x)$$

### Die klassische Lesart

*"Finde eine Funktion $x(t)$, deren Ableitung gleich $f(x(t))$ ist."*

### Die geometrische Lesart

*"An jedem Punkt $x$ des Zustandsraums steht ein Vektor mit Betrag und Richtung $f(x)$."*

Diese Vektoren bilden ein **Vektorfeld**. Eine **Lösung** ist dann eine Kurve, die an jedem Punkt tangential zu den Vektoren verläuft.

Lass uns das visualisieren.

```python
# Beispiel: ẋ = x (exponentielles Wachstum)

def f(x):
    return x

# Zeichne das "Vektorfeld" auf der 1D-Linie
fig, axes = plt.subplots(2, 1, figsize=(12, 8))

# Obere Grafik: f(x) als Funktion
x = np.linspace(-3, 3, 100)
ax1 = axes[0]
ax1.plot(x, f(x), 'b-', linewidth=2, label=r'$f(x) = x$')
ax1.axhline(y=0, color='k', linewidth=0.5)
ax1.axvline(x=0, color='k', linewidth=0.5)
ax1.fill_between(x, f(x), 0, where=(f(x) > 0), alpha=0.2, color='green', label=r'$\dot{x} > 0$: nach rechts')
ax1.fill_between(x, f(x), 0, where=(f(x) < 0), alpha=0.2, color='red', label=r'$\dot{x} < 0$: nach links')
ax1.plot(0, 0, 'ko', markersize=10)
ax1.annotate('Fixpunkt', xy=(0, 0), xytext=(0.5, 1), fontsize=12,
            arrowprops=dict(arrowstyle='->', color='black'))
ax1.set_xlabel(r'$x$')
ax1.set_ylabel(r'$\dot{x} = f(x)$')
ax1.legend(loc='upper left')
ax1.set_title('Die Funktion $f(x)$ bestimmt die Geschwindigkeit')

# Untere Grafik: Das Phasen-Liniendiagramm
ax2 = axes[1]
ax2.axhline(y=0, color='k', linewidth=2)
ax2.set_xlim(-3, 3)
ax2.set_ylim(-0.5, 0.5)

# Pfeile für den Fluss
for xi in np.linspace(-2.5, -0.3, 8):
    ax2.annotate('', xy=(xi - 0.15, 0), xytext=(xi, 0),
                arrowprops=dict(arrowstyle='->', color='red', lw=2))
for xi in np.linspace(0.3, 2.5, 8):
    ax2.annotate('', xy=(xi + 0.15, 0), xytext=(xi, 0),
                arrowprops=dict(arrowstyle='->', color='green', lw=2))

ax2.plot(0, 0, 'ko', markersize=15, fillstyle='none', markeredgewidth=3)
ax2.text(0, -0.25, 'instabil', ha='center', fontsize=12)
ax2.set_xlabel(r'$x$')
ax2.set_yticks([])
ax2.set_title('Phasenlinien-Diagramm: Der Fluss auf der $x$-Achse')

plt.tight_layout()
plt.show()
```

### Beobachtungen

Allein aus dem Vorzeichen von $f(x)$ können wir das gesamte qualitative Verhalten ablesen:

- **Wo $f(x) > 0$**: Das System bewegt sich nach rechts (zu größeren $x$)
- **Wo $f(x) < 0$**: Das System bewegt sich nach links (zu kleineren $x$)
- **Wo $f(x) = 0$**: **Fixpunkte** - das System verharrt

Der Fixpunkt bei $x = 0$ ist hier **instabil**: Jede kleine Störung wird verstärkt. Das System flieht vom Fixpunkt.

*Wir haben das Langzeitverhalten vollständig verstanden, ohne je die Lösung $x(t) = x_0 e^t$ zu berechnen.*

## 1.2 Der Sprung in zwei Dimensionen

In 2D wird die Sache interessanter. Jetzt haben wir:

$$\dot{x} = f(x, y)$$
$$\dot{y} = g(x, y)$$

An jedem Punkt $(x, y)$ gibt es einen Vektor $(f(x,y), g(x,y))$.

Das ergibt ein echtes 2D-Vektorfeld. Die Trajektorien sind nun Kurven in der Ebene.

```python
# Der harmonische Oszillator: ẍ + x = 0
# Als System: ẋ = y, ẏ = -x

def harmonic_oscillator(state):
    x, y = state
    return [y, -x]

# Vektorfeld zeichnen
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Linkes Bild: Vektorfeld
ax1 = axes[0]
x = np.linspace(-3, 3, 20)
y = np.linspace(-3, 3, 20)
X, Y = np.meshgrid(x, y)

U = Y  # ẋ = y
V = -X  # ẏ = -x

# Normalisiere Pfeile für bessere Sichtbarkeit
M = np.sqrt(U**2 + V**2)
M[M == 0] = 1
U_norm = U / M
V_norm = V / M

ax1.quiver(X, Y, U_norm, V_norm, M, cmap='coolwarm', alpha=0.8)
ax1.plot(0, 0, 'ko', markersize=10)
ax1.set_xlabel(r'$x$ (Position)')
ax1.set_ylabel(r'$y = \dot{x}$ (Geschwindigkeit)')
ax1.set_title('Vektorfeld des harmonischen Oszillators')
ax1.set_aspect('equal')
ax1.set_xlim(-3, 3)
ax1.set_ylim(-3, 3)

# Rechtes Bild: Trajektorien
ax2 = axes[1]
ax2.streamplot(X, Y, U, V, density=1.5, color=M, cmap='coolwarm', linewidth=1.5)
ax2.plot(0, 0, 'ko', markersize=10)
ax2.set_xlabel(r'$x$ (Position)')
ax2.set_ylabel(r'$y = \dot{x}$ (Geschwindigkeit)')
ax2.set_title('Trajektorien (Stromlinien)')
ax2.set_aspect('equal')
ax2.set_xlim(-3, 3)
ax2.set_ylim(-3, 3)

plt.tight_layout()
plt.show()
```

### Der Phasenraum

Das rechte Bild zeigt den **Phasenraum** des harmonischen Oszillators.

Die Achsen sind nicht Raum und Zeit, sondern **Position und Geschwindigkeit**. Ein Punkt im Phasenraum repräsentiert den vollständigen *Zustand* des Systems zu einem Zeitpunkt.

Die geschlossenen Kreise bedeuten: **Der Oszillator kehrt immer zu seinem Ausgangszustand zurück** - periodische Bewegung.

Der Ursprung ist ein **Zentrum** - Trajektorien kreisen um ihn, ohne je anzukommen oder zu entkommen.

## 1.3 Existenz und Eindeutigkeit - Die fundamentale Struktur

Ein entscheidender Satz (Picard-Lindelöf, informal):

> Wenn $f$ "hinreichend glatt" ist (Lipschitz-stetig), dann existiert durch jeden Punkt des Phasenraums **genau eine** Trajektorie.

### Konsequenzen

1. **Trajektorien können sich niemals schneiden.** Wenn sie sich in einem Punkt träfen, gäbe es von dort zwei verschiedene Fortsetzungen - Widerspruch zur Eindeutigkeit.

2. **Die Vergangenheit bestimmt die Zukunft vollständig.** Jeder Zustand hat genau eine Geschichte und genau eine Zukunft.

3. **In 2D können Trajektorien sich nicht "überholen".** Sie müssten sich dafür schneiden.

Diese scheinbar abstrakte Eigenschaft hat tiefgreifende Konsequenzen. Sie ist der Grund, warum **in 2D kein Chaos möglich ist** - ein Punkt, auf den wir in späteren Modulen zurückkommen.

```python
# Demonstration: Trajektorien können sich nicht schneiden

def nonlinear_system(state):
    x, y = state
    return [y, -np.sin(x) - 0.3*y]  # Gedämpftes Pendel

fig, ax = plt.subplots(figsize=(12, 8))

# Viele Trajektorien von verschiedenen Startpunkten
t = np.linspace(0, 30, 1000)
colors = plt.cm.viridis(np.linspace(0, 1, 20))

for i, y0 in enumerate(np.linspace(-3, 3, 20)):
    sol = odeint(lambda s, t: nonlinear_system(s), [0, y0], t)
    ax.plot(sol[:, 0], sol[:, 1], color=colors[i], alpha=0.7, linewidth=1)

ax.set_xlabel(r'$\theta$ (Winkel)')
ax.set_ylabel(r'$\dot{\theta}$ (Winkelgeschwindigkeit)')
ax.set_title('Gedämpftes Pendel: Trajektorien nähern sich, schneiden sich aber nie')
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)

plt.show()
```

Beobachte: Alle Trajektorien konvergieren zum Ursprung (dem Ruhezustand des Pendels), aber sie **schneiden sich niemals**, egal wie nahe sie kommen. Sie spiralen gemeinsam einwärts, ohne je zu kollidieren.

## 1.4 Der Fluss als Abbildung

Eine elegante Sichtweise: Das Vektorfeld definiert einen **Fluss** $\phi_t$.

$$\phi_t(x_0) = x(t)$$

wobei $x(t)$ die Lösung ist, die bei $x_0$ startet.

Der Fluss hat die Struktur einer **Gruppe**:
- $\phi_0 = \text{id}$ (Nach null Zeit: kein Fortschritt)
- $\phi_{t+s} = \phi_t \circ \phi_s$ (Nacheinander fließen = Summe der Zeiten)
- $\phi_{-t} = \phi_t^{-1}$ (Zeit umkehren = inverse Abbildung)

Diese Gruppenstruktur spiegelt den **Determinismus** der klassischen Mechanik wider: Die Zeitentwicklung ist umkehrbar, Information geht nicht verloren.

*Für dissipative Systeme (mit Reibung) gilt das nur vorwärts in der Zeit - ein subtiler aber wichtiger Punkt für Chaos.*

```python
# Visualisierung des Flusses: Wie sich ein "Tropfen" im Phasenraum entwickelt

def flow_demo(system, initial_points, times, ax, title):
    """Zeige, wie sich eine Punktmenge unter dem Fluss entwickelt."""
    colors = plt.cm.plasma(np.linspace(0, 1, len(times)))
    
    for i, t_end in enumerate(times):
        t = np.linspace(0, t_end, 100)
        evolved_points = []
        
        for p in initial_points:
            sol = odeint(lambda s, t: system(s), p, t)
            evolved_points.append(sol[-1])
        
        evolved = np.array(evolved_points)
        ax.scatter(evolved[:, 0], evolved[:, 1], c=[colors[i]], s=10, alpha=0.7,
                  label=f't = {t_end:.1f}' if i % 2 == 0 else '')
    
    ax.set_xlabel(r'$x$')
    ax.set_ylabel(r'$y$')
    ax.set_title(title)
    ax.legend()
    ax.set_aspect('equal')

# Erzeuge einen Kreis von Anfangspunkten
theta = np.linspace(0, 2*np.pi, 100)
r = 0.3
initial_circle = np.column_stack([2 + r*np.cos(theta), r*np.sin(theta)])

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# Harmonischer Oszillator: Volumen bleibt erhalten
flow_demo(harmonic_oscillator, initial_circle, [0, 1, 2, 3, 4, 5], axes[0],
          'Harmonischer Oszillator\n(Volumen bleibt erhalten)')

# Gedämpftes System: Volumen schrumpft
def damped_system(state):
    x, y = state
    return [y, -x - 0.5*y]

flow_demo(damped_system, initial_circle, [0, 1, 2, 3, 4, 5], axes[1],
          'Gedämpfter Oszillator\n(Volumen schrumpft - dissipativ)')

plt.tight_layout()
plt.show()
```

### Beobachtung

**Links (Harmonischer Oszillator):** Der Fluss ist **volumenerhaltend**. Der Kreis verformt sich, aber seine Fläche bleibt konstant. Dies ist eine Konsequenz des **Satzes von Liouville** und charakteristisch für Hamiltonsche Systeme.

**Rechts (Gedämpftes System):** Der Fluss ist **kontrahierend**. Volumen schrumpft im Laufe der Zeit. Das System ist **dissipativ** - Energie geht verloren.

Diese Unterscheidung wird fundamental für das Verständnis von Chaos: Seltsame Attraktoren leben in dissipativen Systemen, wo der Fluss Volumen komprimiert, aber trotzdem chaotisch dehnt und faltet.

## 1.5 Autonome vs. Nicht-autonome Systeme

Bisher betrachteten wir **autonome** Systeme: $\dot{x} = f(x)$.

Die rechte Seite hängt nicht explizit von $t$ ab. Das Vektorfeld "steht still".

**Nicht-autonome** Systeme: $\dot{x} = f(x, t)$

Hier ändert sich das Vektorfeld mit der Zeit. Ein erzwungener Oszillator ist ein Beispiel:

$$\ddot{x} + \omega_0^2 x = A\cos(\omega t)$$

### Der Trick: Zeitvariable als Dimension

Jedes nicht-autonome System kann in ein autonomes umgewandelt werden, indem man $t$ als zusätzliche Variable behandelt:

$$\dot{x} = f(x, \tau)$$
$$\dot{\tau} = 1$$

Der Preis: Eine zusätzliche Dimension. Aber das Vektorfeld ist wieder zeitunabhängig.

Für periodische Antriebe kann man $\tau$ auf einen Kreis $S^1$ kompaktifizieren - der Phasenraum wird dann $\mathbb{R}^n \times S^1$, ein Zylinder oder Torus.

## 1.6 Zusammenfassung & Ausblick

### Kernkonzepte dieses Moduls

1. **DGL = Vektorfeld**: Jede DGL definiert an jedem Punkt eine Richtung. Lösungen sind Kurven, die dieser Richtung folgen.

2. **Phasenraum**: Der Raum aller möglichen Zustände. Zeitentwicklung = Bewegung im Phasenraum.

3. **Existenz & Eindeutigkeit**: Durch jeden Punkt verläuft genau eine Trajektorie. Trajektorien schneiden sich nie.

4. **Fluss**: Die Zeitentwicklung als Abbildung $\phi_t$, die Anfangsbedingungen auf Endzustände abbildet.

5. **Dissipativ vs. Konservativ**: Ob der Fluss Volumen erhält oder komprimiert, bestimmt fundamental das mögliche Verhalten.
